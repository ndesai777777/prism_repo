# Funds Dataset CY2024 — Exploratory Data Analysis

Comprehensive variable-by-variable EDA for `Funds_Dataset_CY2024.csv`.

```python
%matplotlib inline
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load the dataset
df = pd.read_csv('Funds_Dataset_CY2024.csv')
print(f'Shape: {df.shape}')
print(f'\nColumns ({len(df.columns)}):')
print(list(df.columns))
```

---
## Numeric Variables

### age

```python
col = 'age'
df[col] = pd.to_numeric(df[col], errors='coerce')
print(f'=== {col} ===')
print(f'Count: {df[col].notna().sum()}')
print(f'Missing: {df[col].isna().sum()} ({df[col].isna().mean()*100:.1f}%)')
print(df[col].describe())
plt.figure(figsize=(8, 3.5))
df[col].dropna().hist(bins=30, edgecolor='white')
plt.title(f'{col} distribution')
plt.xlabel(col)
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```
