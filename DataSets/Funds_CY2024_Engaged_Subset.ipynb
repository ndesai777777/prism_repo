# Funds Dataset CY2024 — Engaged/Intervention Subset

Loads `Funds_Dataset_CY2024.csv` and filters to members who have **any** indication of
program engagement or intervention activity. A member is included if ANY of these fields
are non-missing:

- `Engaged_flag`
- `OptOut_flag`
- `InterventionCount`
- `engaged_date`

```python
import pandas as pd
import numpy as np

# Load full dataset
df = pd.read_csv('Funds_Dataset_CY2024.csv', low_memory=False)
print(f'Full dataset: {df.shape[0]:,} rows, {df.shape[1]} columns')
print(f'Columns: {list(df.columns)}')
```

## Check the engagement/intervention fields

```python
# Check which engagement/intervention columns exist and their missingness
engagement_cols = ['Engaged_flag', 'OptOut_flag', 'InterventionCount', 'engaged_date']
for col in engagement_cols:
    if col in df.columns:
        non_missing = df[col].notna().sum()
        print(f'{col}: {non_missing:,} non-missing ({non_missing/len(df)*100:.1f}%), {df[col].isna().sum():,} missing')
    else:
        print(f'{col}: COLUMN NOT FOUND')
```

## Filter to engaged/intervention members

Include a member if ANY of the four fields above is not missing (i.e., has a value).

```python
# Build inclusion mask: member is included if ANY engagement/intervention field is non-missing
present_engagement_cols = [col for col in engagement_cols if col in df.columns]
print(f'Using columns for filter: {present_engagement_cols}')

inclusion_mask = df[present_engagement_cols].notna().any(axis=1)

df_engaged = df[inclusion_mask].copy().reset_index(drop=True)

print(f'\nFull dataset: {len(df):,} members')
print(f'Engaged/intervention subset: {len(df_engaged):,} members')
print(f'Excluded (no engagement indicators): {len(df) - len(df_engaged):,} members')
print(f'Retention rate: {len(df_engaged)/len(df)*100:.1f}%')
```

## Subset summary

```python
# Show engagement field distributions in the subset
print('=== Engaged Subset: Engagement Field Distributions ===')
print()
for col in present_engagement_cols:
    print(f'--- {col} ---')
    print(df_engaged[col].value_counts(dropna=False).head(10))
    print(f'Non-missing: {df_engaged[col].notna().sum():,}')
    print()
```

```python
# Quick data review of the subset
print(f'Subset shape: {df_engaged.shape}')
print(f'\nSubset dtypes:')
print(df_engaged.dtypes)
print(f'\nSubset first 5 rows:')
df_engaged.head()
```
